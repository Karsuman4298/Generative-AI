```python
prompt = "From: SomeUser@example.com (Some User)\nSubject: Re: baseball"

# Encode the prompt and move to the model's device
inputs = tokenizer(prompt, return_tensors='pt')
input_ids = inputs['input_ids'].to(finetuned_model.device)
attention_mask = inputs['attention_mask'].to(finetuned_model.device)

# Generate text using the finetuned model
output = finetuned_model.generate(
    input_ids,
    attention_mask=attention_mask,
    max_length=200,
    num_return_sequences=1,
    do_sample=True,
    top_k=50,
    top_p=0.95,
    pad_token_id=tokenizer.eos_token_id
)

# Decode and print the generated text
generated_text = tokenizer.decode(output[0], skip_special_tokens=True)
print(generated_text)
```
